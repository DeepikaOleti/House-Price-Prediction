```python
# Create a clean dataset for modeling
# Keep the original df unchanged for EDA

df_model = df.drop(index=extreme_grlivarea.index).copy()

print("Original dataset shape:", df.shape)
print("Modeling dataset shape:", df_model.shape)
print("Rows removed for modeling:", df.shape[0] - df_model.shape[0])
```

```python
# Check missing values in the modeling dataset

missing_model = df_model.isnull().sum()
missing_model = missing_model[missing_model > 0].sort_values(ascending=False)

print("Missing values in modeling dataset:")
print(missing_model)
```

```python
# Handle missing values based on the meaning of each feature

# Categorical columns where missing means the feature is not present
none_cols = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure',
    'BsmtFinType1', 'BsmtFinType2', 'MasVnrType'
]

for col in none_cols:
    df_model[col] = df_model[col].fillna('None')


# Numerical columns where missing means the feature is not present
df_model['MasVnrArea'] = df_model['MasVnrArea'].fillna(0)
df_model['GarageYrBlt'] = df_model['GarageYrBlt'].fillna(0)


# LotFrontage: impute using median LotFrontage of each Neighborhood
df_model['LotFrontage'] = df_model.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)


# Electrical: fill with the most frequent value
df_model['Electrical'] = df_model['Electrical'].fillna(
    df_model['Electrical'].mode()[0]
)

print("Missing values handled.")
```

```python
# Verify remaining missing values

remaining_missing = df_model.isnull().sum()
remaining_missing = remaining_missing[remaining_missing > 0]

print("Remaining missing values:")
print(remaining_missing)

print("\nTotal remaining missing values:", df_model.isnull().sum().sum())
```

```python
# Feature Engineering

# Total area of the house
df_model['TotalSF'] = (
    df_model['TotalBsmtSF'] +
    df_model['1stFlrSF'] +
    df_model['2ndFlrSF']
)

# Total number of bathrooms
df_model['TotalBathrooms'] = (
    df_model['FullBath'] +
    0.5 * df_model['HalfBath'] +
    df_model['BsmtFullBath'] +
    0.5 * df_model['BsmtHalfBath']
)

# Total porch/deck area
df_model['TotalPorchSF'] = (
    df_model['OpenPorchSF'] +
    df_model['3SsnPorch'] +
    df_model['EnclosedPorch'] +
    df_model['ScreenPorch'] +
    df_model['WoodDeckSF']
)

# Age of the house at the time of sale
df_model['HouseAgeAtSale'] = df_model['YrSold'] - df_model['YearBuilt']

# Years since the last remodeling
df_model['YearsSinceRemodel'] = df_model['YrSold'] - df_model['YearRemodAdd']

# Display the newly created features
df_model[
    ['TotalSF', 'TotalBathrooms', 'TotalPorchSF',
     'HouseAgeAtSale', 'YearsSinceRemodel']
].head()
```

```python
# Separate features and target variable

X = df_model.drop(columns=['SalePrice'])
y = df_model['SalePrice']

print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)
```

```python
# Encode categorical variables using one-hot encoding

X_encoded = pd.get_dummies(X, drop_first=True)

print("Original feature shape:", X.shape)
print("Encoded feature shape:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Check target skewness before modeling

print("Original SalePrice skewness:", y_train.skew())

# Log-transform the target to reduce right skewness
y_train_log = np.log1p(y_train)

print("Log-transformed SalePrice skewness:", y_train_log.skew())
```

```python
# Train a baseline Linear Regression model

from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train, y_train_log)

# Predict on test data
y_pred_log = lr_model.predict(X_test)

print("Linear Regression model trained successfully.")
```

```python
# Convert log predictions back to original SalePrice scale

y_pred = np.expm1(y_pred_log)

print("Predictions converted back to original SalePrice scale.")
print("First 5 predictions:")
print(y_pred[:5])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Model Performance:")
print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
# Train Random Forest Regression model

from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train_log)

print("Random Forest model trained successfully.")
```

```python
# Predict on test data
y_rf_pred_log = rf_model.predict(X_test)

# Convert predictions back to original SalePrice scale
y_rf_pred = np.expm1(y_rf_pred_log)

print("Random Forest predictions converted back to original SalePrice scale.")
print("First 5 predictions:")
print(y_rf_pred[:5])
```

```python
# Evaluate Random Forest model

rf_mae = mean_absolute_error(y_test, y_rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, y_rf_pred))
rf_r2 = r2_score(y_test, y_rf_pred)

print("Random Forest Model Performance:")
print("MAE:", rf_mae)
print("RMSE:", rf_rmse)
print("R² Score:", rf_r2)
```

```python
# Train Gradient Boosting Regression model

from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_model.fit(X_train, y_train_log)

print("Gradient Boosting model trained successfully.")
```

```python
# Predict on test data using Gradient Boosting

y_gb_pred_log = gb_model.predict(X_test)

# Convert predictions back to original SalePrice scale
y_gb_pred = np.expm1(y_gb_pred_log)

print("Gradient Boosting predictions converted back to original SalePrice scale.")
print("First 5 predictions:")
print(y_gb_pred[:5])
```

```python
# Evaluate Gradient Boosting model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

gb_mae = mean_absolute_error(y_test, y_gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, y_gb_pred))
gb_r2 = r2_score(y_test, y_gb_pred)

print("Gradient Boosting Model Performance:")
print("MAE:", gb_mae)
print("RMSE:", gb_rmse)
print("R² Score:", gb_r2)
```

```python
# Compare model performance

model_comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "Gradient Boosting"],
    "MAE": [mae, rf_mae, gb_mae],
    "RMSE": [rmse, rf_rmse, gb_rmse],
    "R2 Score": [r2, rf_r2, gb_r2]
})

print("Model Performance Comparison:")
display(model_comparison)
```

```python
import matplotlib.pyplot as plt

# Plot R² Score comparison
plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["Model"],
    model_comparison["R2 Score"]
)

plt.title("Model Comparison - R² Score")
plt.xlabel("Model")
plt.ylabel("R² Score")
plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```

# Challenges Faced and Techniques Used

## 1. Missing Values

### Challenge
The dataset contained missing values in several features. Some columns had a very high percentage of missing values, such as PoolQC, MiscFeature, Alley, Fence and FireplaceQu.

### Technique Used
Missing values were handled according to the meaning of the corresponding feature. For categorical features where missing indicated the absence of the feature, the value was replaced with `None`. Numerical features representing absent facilities were replaced with 0. `LotFrontage` was imputed using the median value within each neighborhood, while the missing `Electrical` value was replaced with the mode.

### Reasoning
A single imputation strategy for all columns could introduce incorrect information. For example, a missing PoolQC value can indicate that the property does not have a pool rather than an unknown pool quality.

---

## 2. Skewed Target Variable

### Challenge
The `SalePrice` variable was positively skewed.

### Technique Used
The target variable was transformed using the natural logarithm through `np.log1p()` before model training.

### Reasoning
The transformation reduced the skewness of the training target from approximately 2.05 to 0.23, making the target distribution more suitable for regression modeling.

---

## 3. Extreme Observations

### Challenge
The `GrLivArea` analysis identified unusually large living-area observations. Two extreme observations had very large living areas combined with comparatively low sale prices.

### Technique Used
The two identified extreme observations were excluded from the modeling dataset while the original dataset was retained for exploratory analysis.

### Reasoning
These observations could disproportionately influence regression models. Keeping the original dataset unchanged also allows the exploratory analysis to remain reproducible.

---

## 4. Categorical Variables

### Challenge
The dataset contained 43 categorical variables that could not be directly used by most machine learning algorithms.

### Technique Used
